```python

#plt.plot()

import matplotlib.pyplot as plt

x = [1,2,3,4,5]
y = [2,3,5,7,11]

plt.plot(x, y)
plt.title("Sample Plot for practice purpose   ")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.show()
```

```python
#plt.scatter()
import matplotlib.pyplot as plt

x = [1,2,3,4,5]
y = [2,3,5,7,11]

plt.scatter(x, y)
plt.title("Sample Scatter Plot for practice purpose   ")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.show()
```

```python
#plt.bar()

x =['Apple', 'Banana', 'Cherry', 'Date', 'Elderberry']
y =[20, 35, 30, 35, 27]
plt.bar(x, y, color='blue')
plt.title("sales on this month")
plt.xlabel("fruits")
plt.ylabel("quantity of fruits")
plt.show()
```

```python
#plt.barh()

x = ['vijay', 'ajith', 'dhanush']
y = [10, 9.3, 8]

plt.barh(x, y, color='skyblue')
plt.title("actor rating and Popularity")
plt.xlabel("actors name")
plt.ylabel("rating and popularity")
plt.show()
```

```python
#plt.hist()

import numpy as np

data = np.random.randn(1000)

plt.hist(data, bins=20, color='green', edgecolor='black')
plt.title("Histogram Example")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.show()
```

```python
#plt.pie()

sizes =[21,22,24]
labels =['pawan','nithin','kethan']
plt.pie(sizes, labels=labels ,autopct='%1.1f%%', startangle=140 )
plt.title("basic_information")
plt.show()
```

```python
#plt.boxplot()

data = [5, 8, 10, 15, 18, 20, 22, 25, 30]
plt.boxplot(data)
plt.title("Box Plot Example")
plt.ylabel("Values")
plt.show()
```

```python
#plt.Subplot alows multiple plots

x = [1, 2, 3, 4]
y1 = [2, 4, 6, 8]
y2 = [1, 3, 5, 7]

plt.subplot(1, 2, 1)
plt.plot(x, y1)
plt.title("Plot 1")

plt.subplot(1, 2, 2)
plt.bar(x, y2)
plt.title("Plot 2")

plt.show()
```

```python

#plt.legend()

x = [1, 2, 3, 4]
plt.plot(x, [1, 4, 9, 16], label='Square')
plt.plot(x, [1, 8, 27, 64], label='Cube')

plt.legend()
plt.title("Legend Example")
plt.show()
```

```python
#plt.grid()

x = [1, 2, 3, 4]
y = [2, 4, 6, 8]

plt.plot(x, y)
plt.grid(True)
plt.title("Line Plot with Grid")
plt.show()
```

```python
#plt.figure()

plt.Figure(figsize=(3,3))
plt.plot([1,1,2],[44,4,55])
plt.title("custom figure size")
plt.show()
```

```python
plt.plot([1, 2, 3], [2, 4, 6])
plt.title("Save Plot Example")
plt.savefig("plot_output.png")
plt.show()
```

```python
#plt.errorbar()

x =[1,2,3,4]
y =[2,4,6,8]
error =[0.2,0.3,0.5,0.4]
plt.errorbar(x,y,yerr=error,fmt='o',color='black',ecolor='red',capsize=5)
plt.title("error bar")
plt.show()
```

```python
#plt.fill_between()
import numpy as np

x= np.linspace(0,10,100)
y = np.sin(x)
plt.plot(x,y)
plt.fill_between(x,y,color='blue',alpha=0.4)
plt.title("fill_between examples")
plt.show()
```

```python
#plt.annotate()

x =[1,2,3,4,5]
y =[2,4,6,8,10]
plt.plot(x,y)
plt.annotate("Peak point", xy=(5,10),xytext=(3,9),
             arrowprops=dict(facecolor='black',shrink=(0.5)))
plt.show()
```
